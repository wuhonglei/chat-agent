```python
from pydantic import BaseModel, Field


class UserModel(BaseModel):
    # Python 使用 snake_case，但 API 可能返回 camelCase
    user_name: str = Field(alias="userName")
    first_name: str = Field(alias="firstName")
    last_name: str = Field(alias="lastName")
    email_address: str = Field(alias="emailAddress")
    is_active: bool = Field(alias="isActive")
    age: int = Field(..., description='person age')


# 接收 camelCase 的 JSON 数据
json_data = {
    "userName": "john_doe",
    "firstName": "John",
    "lastName": "Doe",
    "emailAddress": "john@example.com",
    "isActive": True,
    "age": 18
}

# Pydantic 自动映射到 Python 的 snake_case
user = UserModel(**json_data)
print(user.user_name)  # "john_doe"
print(user.first_name)  # "John"

# 输出时也可以使用别名
print(user.model_dump(by_alias=True))
# {"userName": "john_doe", "firstName": "John", ...}
```

```python
from pydantic import BaseModel, Field
from typing import Optional


class Meta(BaseModel):
    age: Optional[int] = Field(default=None, description="person age")


meta = Meta()
print(meta.age)

# 测试验证
print("测试验证:")
try:
    meta2 = Meta(age=25)
    print(f"age=25: {meta2.age}")
except Exception as e:
    print(f"错误: {e}")

try:
    meta3 = Meta(age=None)
    print(f"age=None: {meta3.age}")
except Exception as e:
    print(f"错误: {e}")
```
